# 模型初始化的参数使用情况

## 1、temperature的使用
举例1

```python
import os

from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

# 读取.env配置文件中的信息。相关环境变量以.env文件优先
load_dotenv(override=True)

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
OPENAI_API_BASE = os.getenv("OPENAI_API_BASE")

# 2、模型初始化
llm_deepseek = init_chat_model(
    model="openai:deepseek-v4-pro",
    temperature=0,
    api_key=OPENAI_API_KEY,
    base_url=OPENAI_API_BASE,
)

# 模型调用
for i in range(3):
    print(llm_deepseek.invoke("帮我写一首关于春天的七言绝句").content)
```

举例2：应用

```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
import os
# 从.env文件中加载环境变量
load_dotenv(override=True)
model = init_chat_model(
    model="openai:deepseek-v4-pro",
    temperature=0,
    api_key=os.getenv("OPENAI_API_KEY"),
    base_url=os.getenv("OPENAI_API_BASE"),
)
# 向模型发送单条数据
response = model.invoke("张三，男，30岁，拥有8年编程开发经验，目前在某互联网大厂担任技术专家。帮我从上文中提取数据，返回JSON格式")
print(response.content)
```

举例3：应用

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os
# 从.env文件中加载环境变量
load_dotenv(override=True)
model = init_chat_model(
    model="openai:deepseek-v4-pro",
    temperature=1.5,
    api_key=os.getenv("OPENAI_API_KEY"),
    base_url=os.getenv("OPENAI_API_BASE"),
)
# 向模型发送单条数据
response = model.invoke("请为一款极致静音的机械键盘写3个充满诗意且极具张力的广告语。")
print(response.content)
```

## 2、max_tokens参数的使用

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os
# 从.env文件中加载环境变量
load_dotenv(override=True)
model = init_chat_model(
    model="openai:deepseek-v4-pro",
    # temperature=1.5,
    api_key=os.getenv("OPENAI_API_KEY"),
    base_url=os.getenv("OPENAI_API_BASE"),
    max_tokens=5,
    extra_body={
        "thinking": {"type": "disabled"}
    }
)
# 向模型发送单条数据
response = model.invoke("介绍一下你自己")
print(response)
```
